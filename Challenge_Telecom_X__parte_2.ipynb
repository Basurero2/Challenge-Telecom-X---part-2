##**Eliminación de columnas irrelevantes**

```python
datos.columns
```

```python
datos.drop(columns=['Cliente_ID','Genero','Pareja','Dependientes','Servicio_telefonico','Varias_lineas','Servicio_internet','Copia_seguridad_en_linea','Proteccion_dispositivo','Televisión_por_cable','Streaming_peliculas','Factura_en_linea','Cuentas_diarias', 'N_servicios'], axis=1, inplace=True)
datos.info()
```

##**Separación de Datos**

```python
X = datos.drop('Abandono', axis=True)
y = datos.Abandono

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3, stratify=y, random_state=123)
```

## **Encoding**

```python
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import OneHotEncoder
```

```python
columnas = X.columns
```

```python
one_hot = make_column_transformer((OneHotEncoder(drop='if_binary', handle_unknown='ignore'),
                                   ['Tipo_servicio_internet', 'Tipo_contrato', 'Forma_pago']),
                                  remainder='passthrough',
                                  sparse_threshold=0,
                                  force_int_remainder_cols=False)
```

```python
one_hot.fit(X_train)
```

```python
X_train_transf = one_hot.transform(X_train)
X_test_transf = one_hot.transform(X_test)
```

##**Verificación de la Proporción de Cancelación (Churn)**

```python
datos['Abandono'].value_counts(normalize=True).round(2)
```

##**Balanceo de Clases**

```python
from imblearn.over_sampling import SMOTE
```

```python
oversampling = SMOTE(random_state=123)
X_train_bal, y_train_bal = oversampling.fit_resample(X_train_transf,y_train)
y_train_bal.value_counts(normalize=True)
```

##**Normalizando los datos**

La **normalización** de los datos garantiza que todas las variables contribuyan en igualdad de condiciones al proceso de modelado. En modelos basados en distancia, como **KNN**, la magnitud de las variables influye directamente en el cálculo de la distancia entre observaciones. Si las variables no están escaladas, aquellas con mayor rango numérico pueden dominar la métrica de distancia e introducir un sesgo en la clasificación.

Al aplicar una normalización (por ejemplo, mediante **MinMaxScaler**), todas las variables se reescalan a un mismo rango —habitualmente entre 0 y 1—, lo que evita que aquellas con valores numéricos más altos dominen el proceso de aprendizaje. De esta manera, la magnitud original de cada variable no determina su influencia en modelos basados en distancia (como **KNN**) o en optimización de parámetros (como la **regresión logística** o las **redes neuronales**).

En contraste, los **modelos basados en árboles no requieren normalización**, ya que realizan particiones mediante umbrales y no dependen de métricas de distancia.

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
normalizacion = MinMaxScaler()
X_train_bal_normalizado = normalizacion.fit_transform(X_train_bal)
X_test_normalizado = normalizacion.transform(X_test_transf)
```

```python
pd.DataFrame(X_train_bal_normalizado)
```

#**Creación de Modelos**

##**Baseline**

```python
from sklearn.dummy import DummyClassifier
```

```python
dummy = DummyClassifier(strategy='stratified')
```

```python
dummy.fit(X_train_transf, y_train)
```

```python
dummy.score(X_test, y_test)
```

##**Árbol de decisión**

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
modelo_arbol = DecisionTreeClassifier(max_depth=4, min_samples_split=2, min_samples_leaf=1, random_state=123)
modelo_arbol.fit(X_train_bal, y_train_bal)
```

```python
modelo_arbol.score(X_test_transf, y_test)
```

##**KNN**

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors=15, weights='distance')
```

```python
knn.fit(X_train_bal_normalizado, y_train_bal)
```

```python
knn.score(X_test_normalizado, y_test)
```

##**Evaluación de los Modelos**

```python
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score

y_dummy_predict = dummy.predict(X_test_transf)
y_arbol_predict = modelo_arbol.predict(X_test_transf)
y_knn_predict = knn.predict(X_test_normalizado)


resultados = (pd.DataFrame({
    'Modelo' : ['Dummy', 'Decision Tree', 'KNN'],
    'Accuracy':[
        accuracy_score(y_test, y_dummy_predict),
        accuracy_score(y_test, y_arbol_predict),
        accuracy_score(y_test, y_knn_predict)
    ],
    'Recall' : [
        recall_score(y_test, y_dummy_predict),
        recall_score(y_test, y_arbol_predict),
        recall_score(y_test, y_knn_predict)
    ],
    'Precision' : [
        precision_score(y_test, y_dummy_predict),
        precision_score(y_test, y_arbol_predict),
        precision_score(y_test, y_knn_predict),
    ],
    'F1': [
        f1_score(y_test, y_dummy_predict),
        f1_score(y_test, y_arbol_predict),
        f1_score(y_test, y_knn_predict)
    ]
})).round(2)


resultados
```

#**Análisis de la Importancia de las Variables**

##**KNN**

```python
from sklearn.inspection import permutation_importance

resultados = permutation_importance(
    knn,
    X_test_normalizado,
    y_test,
    n_repeats=10,
    random_state=123,
    scoring='recall'
)

nombre_columnas = one_hot.get_feature_names_out()

importancia_knn = pd.Series(resultados.importances_mean, index=nombre_columnas)
importancia_knn.sort_values(ascending=False)
```

##**Modelo Árbol**

```python
importancia_arbol = pd.Series(
    modelo_arbol.feature_importances_,
    index=nombre_columnas
)

importancia_arbol.sort_values(ascending=False)
```

```python
resultados_arbol = permutation_importance(
    modelo_arbol,
    X_test_transf,
    y_test,
    n_repeats=10,
    random_state=123,
    scoring='recall'
)

importancia_arbol = pd.Series(
    resultados_arbol.importances_mean,
    index=nombre_columnas
)

importancia_arbol.sort_values(ascending=False)
```

## 🚩 Factores Determinantes del Abandono (Churn)



Tras la validación del modelo, se han identificado cuatro pilares que explican por qué los clientes deciden finalizar su contrato con la compañía:

---

### 📅 1. Contrato Mensual (Month-to-month)
* **Impacto:** Es el principal predictor de fuga.
* **Análisis:** Los clientes sin compromiso de permanencia presentan una probabilidad significativamente mayor de cancelar. Esto confirma que los contratos de corto plazo carecen de **barreras de salida**, facilitando la migración hacia la competencia.

### 💳 2. Pago mediante Electronic Check
* **Impacto:** Correlación positiva con la deserción.
* **Análisis:** Los medios de pago **poco automatizados** o que requieren una acción manual del cliente están asociados a una mayor tasa de cancelación, posiblemente por la fricción constante en el proceso de facturación mensual.

### 🌐 3. Servicio de Internet por Fibra Óptica
* **Impacto:** Segmento de alto riesgo.
* **Análisis:** Aunque es una tecnología superior, muestra una asociación positiva con el churn. Esto indica un perfil de cliente **más exigente**, sensible a variaciones en la calidad del servicio y a los costos más elevados del plan premium.



### ⏳ 4. Baja Antigüedad (Meses_permanencia)
* **Impacto:** Factor de riesgo temporal.
* **Análisis:** El riesgo de abandono es crítico durante los **primeros meses de servicio**. A medida que aumenta la permanencia, el cliente genera mayor fidelidad, lo que reduce drásticamente la probabilidad de salida.

---

> **Conclusión de Valor:** La combinación de un contrato mensual con tecnología de fibra óptica y pago manual crea el "perfil de riesgo máximo". Las campañas de retención deben priorizar la conversión de estos usuarios hacia métodos de pago automáticos y contratos de larga duración.

## 🎯 Plan de Acción: Estrategias de Retención



Para reducir el impacto del **Churn**, el enfoque debe priorizar la conversión de clientes de alto riesgo en usuarios fidelizados mediante las siguientes tácticas:

---

### 1. Conversión de Contratos
* **Estrategia:** Incentivar la migración de contratos "Mes a Mes" hacia acuerdos anuales o bianuales.
* **Táctica:** Ofrecer descuentos exclusivos o bonificaciones en servicios adicionales como recompensa por el compromiso de permanencia.

### 2. Blindaje de los Primeros 6 Meses
* **Estrategia:** Implementar un **Programa de Fidelización Temprana** para nuevos usuarios.
* **Táctica:** Realizar llamadas proactivas de soporte y encuestas de satisfacción para ayudar al cliente a superar la "zona crítica" de mayor probabilidad de abandono.



### 3. Valor Agregado en Fibra Óptica
* **Estrategia:** Fortalecer la percepción de calidad en el segmento de tecnología Premium.
* **Táctica:** Incluir servicios complementarios como **Soporte Técnico** o **Seguridad Digital** sin costo extra para justificar el precio y retener a los clientes más exigentes.

### 4. Automatización de Pagos
* **Estrategia:** Reducir la fricción financiera asociada al pago manual.
* **Táctica:** Promover activamente el **Débito Automático** (Cargo a cuenta) mediante beneficios únicos o pequeñas rebajas porcentuales en la factura mensual.

---

> **💡 Foco Estratégico:** La prioridad absoluta debe ser el cruce de variables: usuarios de **Fibra Óptica** con **pago manual**, ya que representan el perfil con mayor riesgo de fuga inmediata según nuestro modelo.
